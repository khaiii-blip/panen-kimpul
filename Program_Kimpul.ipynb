```python
# Sistem Hasil Panen Kimpul

def input_data():
    nama_petani = input("Masukkan uang: ")
    jumlah_panen = float(input("Masukkan jumlah panen kimpul (kg): "))
    harga_per_kg = float(input("Masukkan harga kimpul per kg: "))

    return nama_petani, jumlah_panen, harga_per_kg


def buat_laporan(nama_petani, jumlah_panen, harga_per_kg):
    total = jumlah_panen * harga_per_kg

    print("\n================================")
    print("     LAPORAN PANEN KIMPUL")
    print("================================")
    print("Nama petani  :", nama_petani)
    print("Jumlah panen :", jumlah_panen, "kg")
    print("Harga per kg : Rp", harga_per_kg)
    print("Total hasil  : Rp", total)


# Program utama
nama_petani, jumlah_panen, harga_per_kg = input_data()
buat_laporan(nama_petani, jumlah_panen, harga_per_kg)
```
